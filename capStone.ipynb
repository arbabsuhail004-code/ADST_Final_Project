```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\ADMIN\Downloads\capstoneProject\amazon.csv")
```

```python
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("\nDataset Information:")
df.info()
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
print(df[['discounted_price', 'actual_price']].head(10))
```

```python
df['discounted_price'] = df['discounted_price'].str.replace('₹', '', regex=False)
df['discounted_price'] = df['discounted_price'].str.replace(',', '', regex=False)
```

```python
df['actual_price'] = df['actual_price'].str.replace('₹', '', regex=False)
df['actual_price'] = df['actual_price'].str.replace(',', '', regex=False)
```

```python
df['discounted_price'] = pd.to_numeric(df['discounted_price'], errors='coerce')
df['actual_price'] = pd.to_numeric(df['actual_price'], errors='coerce')
```

```python
print(df[['discounted_price', 'actual_price']].dtypes)
display(df[['discounted_price', 'actual_price']].head())
```

```python
df['discount_percentage'] = df['discount_percentage'].str.replace('%', '', regex=False)
```

```python
df['discount_percentage'] = pd.to_numeric(
    df['discount_percentage'], errors='coerce'
)
```

```python
display(df[['discount_percentage']].head())
print(df['discount_percentage'].dtype)
```

```python
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
```

```python
df['rating_count'] = df['rating_count'].str.replace(',', '', regex=False)
```

```python
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce')
```

```python
display(df[['rating', 'rating_count']].head(10))
```

```python
print("\nMissing values after conversion:")
print(df[['rating', 'rating_count']].isnull().sum())
```

```python

print("Rating values outside the 0–5 range:")
display(df[(df['rating'] < 0) | (df['rating'] > 5)])

print("Negative rating counts:")
display(df[df['rating_count'] < 0])

print("Negative prices:")
display(df[(df['discounted_price'] < 0) | (df['actual_price'] < 0)])
```

```python
print("Unique product IDs:", df['product_id'].nunique())

print("Repeated product IDs:", df['product_id'].duplicated().sum())
```

```python
duplicate_products = df[
    df['product_id'].duplicated(keep=False)
]

# Display repeated products
display(
    duplicate_products[
        ['product_id', 'product_name', 'category',
         'discounted_price', 'actual_price', 'rating']
    ].sort_values('product_id').head(20)
)
```

```python
print("Missing ratings:")
display(df[df['rating'].isnull()])

print("Missing rating counts:")
display(df[df['rating_count'].isnull()])
```

```python

# Check blank values in text columns
for col in df.select_dtypes(include=['object', 'str']).columns:
    count = df[col].apply(
        lambda x: isinstance(x, str) and x.strip() == ''
    ).sum()

    if count > 0:
        print(col, ":", count, "blank values")
```

```python
print("Missing values before cleaning:")
print(df.isnull().sum())

# Remove rows where rating or rating_count is missing
df = df.dropna(subset=['rating', 'rating_count'])

# Check missing values again
print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python

# Check the final number of rows and columns
print("Dataset Shape:", df.shape)

# Check data types
print("\nData Types:")
print(df.dtypes)

# Check numeric column statistics
print("\nSummary Statistics:")
display(
    df[['discounted_price', 'actual_price',
        'discount_percentage', 'rating',
        'rating_count']].describe()
)
```

```python
print("Missing values remaining:", df.isnull().sum().sum())
print("Duplicate rows remaining:", df.duplicated().sum())
```

```python
df.to_csv("amazon_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
print("Ratings below 0 or above 5:",
      ((df['rating'] < 0) | (df['rating'] > 5)).sum())

# Check negative prices
print("Negative prices:",
      ((df['discounted_price'] < 0) |
       (df['actual_price'] < 0)).sum())

# Check invalid discount percentages
print("Invalid discounts:",
      ((df['discount_percentage'] < 0) |
       (df['discount_percentage'] > 100)).sum())

# Check negative rating counts
print("Negative rating counts:",
      (df['rating_count'] < 0).sum())
```

```python
display(
    df[['discounted_price', 'actual_price',
        'discount_percentage', 'rating',
        'rating_count']].describe().round(2)
)
```

```python
category_rating = df.groupby('category')['rating'].mean()

category_rating = category_rating.sort_values(ascending=False)

display(category_rating.round(2))
```

```python

import matplotlib.pyplot as plt

category_rating.sort_values().plot(
    kind='barh',
    figsize=(10, 6),
    title='Average Product Rating by Category'
)

plt.xlabel('Average Rating')
plt.ylabel('Category')
plt.tight_layout()
plt.show()
```
